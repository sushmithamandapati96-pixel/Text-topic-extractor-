```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
documents = [
    "cat dog animal",
    "dog cat pet",
    "car bus vehicle",
    "bus car transport"
]

print("Documents:")
for i, doc in enumerate(documents, 1):
    print(i, ":", doc)
```

```python
words = sorted(set(word for doc in documents for word in doc.split()))

matrix = np.array([
    [doc.split().count(word) for word in words]
    for doc in documents
])

print("Words:", words)
print("\nDocument × Word Matrix:")
print(matrix)
```

```python
U, S, VT = np.linalg.svd(matrix, full_matrices=False)

print("Singular values:")
print(S)
```

```python
number_of_topics = 2

topic_word_scores = VT[:number_of_topics]

print("Hidden Topics:\n")

for topic_number, scores in enumerate(topic_word_scores, 1):
    important_words = sorted(
        zip(words, scores),
        key=lambda x: abs(x[1]),
        reverse=True
    )

    print("Topic", topic_number)
    print(important_words[:4])
    print()
```
